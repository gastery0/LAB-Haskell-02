```haskell
-- 1. Procesar lecturas
procesarLecturas :: (Double -> Double) -> [Double] -> [Double]
procesarLecturas transformacion = map transformacion

-- 2. Filtro personalizado
crearFiltro :: (a -> Bool) -> ([a] -> [a])
crearFiltro condicion = filter condicion

-- 3. Verificadores de rangos
enRango :: Int -> Int -> (Int -> Bool)
enRango minVal maxVal = \x -> x >= minVal && x <= maxVal

-- 4. Potenciador
potenciador :: Int -> (Int -> Int)
potenciador p = \x -> x ^ p
```

```haskell
-- Pruebas Ejercicio 4
let celsiusAFaranheit c = (c * 9/5) + 32
print $ procesarLecturas celsiusAFaranheit [0,12,20]

let soloPares = crearFiltro even
print $ soloPares [1..10]

let entre5y10 = enRango 5 10
print $ entre5y10 7
print $ entre5y10 12

let potencia5 = potenciador 5
print $ potencia5 2
```

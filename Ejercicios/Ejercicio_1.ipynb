```haskell
-- 1. Funciones usando recursión
miLength :: [a] -> Int
miLength [] = 0
miLength (_:xs) = 1 + miLength xs

miLast :: [a] -> a
miLast [x] = x
miLast (_:xs) = miLast xs

duplicar :: [Int] -> [Int]
duplicar [] = []
duplicar (x:xs) = (x * 2) : duplicar xs

eliminarPares :: [Int] -> [Int]
eliminarPares [] = []
eliminarPares (x:xs)
  | odd x     = x : eliminarPares xs
  | otherwise = eliminarPares xs

-- 2. Funciones usando orden superior
miLengthHOF :: [a] -> Int
miLengthHOF = foldr (\_ n -> n + 1) 0

miLastHOF :: [a] -> a
miLastHOF = foldl1 (\_ x -> x)

duplicarHOF :: [Int] -> [Int]
duplicarHOF = map (* 2)

eliminarParesHOF :: [Int] -> [Int]
eliminarParesHOF = filter odd
```

```haskell
-- Pruebas Ejercicio 1
print $ miLength [1,2,3,4,5]
print $ miLast [10, 20, 30]
print $ duplicar [1, 2, 3]
print $ eliminarPares [1, 2, 3, 4]
print $ miLengthHOF [1..10]
print $ duplicarHOF [5, 10]
print $ eliminarParesHOF [1, 2, 3, 4]
```

```haskell
import Data.Char (isUpper)

-- 1. Reescribir sin recursión explícita
sumaLista :: [Int] -> Int
sumaLista = foldr (+) 0

mayoresQue10 :: [Int] -> Int
mayoresQue10 = length . filter (> 10)

productoLista :: [Int] -> Int
productoLista = foldr (*) 1

celsiusAFahrenheit :: [Double] -> [Double]
celsiusAFahrenheit = map (\c -> (c * 9/5) + 32)

ajustarImpares :: [Int] -> [Int]
ajustarImpares = map (\x -> if odd x then x + 5 else x)

-- 2. Usando filter
nombresLargos :: [String] -> [String]
nombresLargos = filter (\n -> length n > 5)

soloNegativos :: [Int] -> [Int]
soloNegativos = filter (< 0)

edadesUniversitarios :: [Int] -> [Int]
edadesUniversitarios = filter (\e -> e >= 18 && e <= 25)

-- 3. Usando filter y map
agregarIVA :: [Double] -> [Double]
agregarIVA = map (* 1.19) . filter (>= 1000)

longitudesVocales :: [String] -> [Int]
longitudesVocales = map length . filter (\w -> not (null w) && head w `elem` "aeiouAEIOU")

cuadradosPares :: [Int] -> [Int]
cuadradosPares = map (^2) . filter even

-- 4. Usando any o all
hayNegativo :: [Int] -> Bool
hayNegativo = any (< 0)

todosPares :: [Int] -> Bool
todosPares = all even

nombresCorrectos :: [String] -> Bool
nombresCorrectos = all (\w -> not (null w) && isUpper (head w))

-- 5. Usando takeWhile o dropWhile
menoresQue100 :: [Int] -> [Int]
menoresQue100 = takeWhile (< 100)

sinCerosIniciales :: [Int] -> [Int]
sinCerosIniciales = dropWhile (== 0)
```

```haskell
-- Pruebas Ejercicio 3
print $ sumaLista [1,2,3,4]
print $ celsiusAFahrenheit [0, 100]
print $ ajustarImpares [1, 2, 3, 4]
print $ nombresLargos ["Juan", "Sebastián"]
print $ agregarIVA [500, 1000, 2000]
print $ longitudesVocales ["arbol", "luna"]
print $ nombresCorrectos ["Ana", "Pedro"]
print $ sinCerosIniciales [0, 0, 1, 0, 2]
```

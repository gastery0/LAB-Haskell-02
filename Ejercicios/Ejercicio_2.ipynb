```haskell
-- 1. Comprensión de listas básicas
cuadrados :: [Int]
cuadrados = [x^2 | x <- [1..10]]

divisiblesPor3 :: [Int]
divisiblesPor3 = [x | x <- [1..30], x `mod` 3 == 0]

paresMenores :: [(Int, Int)]
paresMenores = [(x, y) | x <- [1..5], y <- [1..5], x < y]

-- 2. Divisores de un número
divisores :: Int -> [Int]
divisores n = [x | x <- [1..n], n `mod` x == 0]

-- 3. Número perfecto
divisoresPropios :: Int -> [Int]
divisoresPropios n = [x | x <- [1..(n-1)], n `mod` x == 0]

esPerfecto :: Int -> Bool
esPerfecto n = sum (divisoresPropios n) == n
```

```haskell
-- Pruebas Ejercicio 2
print cuadrados
print divisiblesPor3
print paresMenores
print $ divisores 12
print $ esPerfecto 28
print $ esPerfecto 10
```

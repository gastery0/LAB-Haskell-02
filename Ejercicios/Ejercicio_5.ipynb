```haskell
import Data.List (sortOn)

type Estado = (Int, Int)
type Nodo = Estado
type Costo = Int
type Grafo = [(Nodo, [(Nodo, Costo)])]

capA, capB :: Int
capA = 5
capB = 3

-- Transferir de Jarra A a Jarra B
transferirAB :: Estado -> Estado
transferirAB (a, b) =
  let espacioDisponibleB = capB - b
      cantidadAVerter = min a espacioDisponibleB
  in (a - cantidadAVerter, b + cantidadAVerter)

-- Transferir de Jarra B a Jarra A
transferirBA :: Estado -> Estado
transferirBA (a, b) =
  let espacioDisponibleA = capA - a
      cantidadAVerter = min b espacioDisponibleA
  in (a + cantidadAVerter, b - cantidadAVerter)

-- Generador de estados vecinos
vecinos :: Estado -> [(Estado, Costo)]
vecinos (a,b) =
  [ ((capA, b), 1)
  , ((a, capB), 1)
  , ((0, b), 1)
  , ((a, 0), 1)
  , (transferirAB (a,b), 1)
  , (transferirBA (a,b), 1)
  ]

-- Heurística admisible
heuristica :: Estado -> Costo
heuristica (a, _) = abs (4 - a)

-- Algoritmo A*
aStar :: (Nodo -> [(Nodo, Costo)]) -> (Nodo -> Costo) -> Nodo -> Nodo -> [(Nodo, Costo)]
aStar sucesores heur_fn inicio meta = buscar [(inicio, 0)] []
  where
    buscar [] _ = []
    buscar ((nodo, costo):cola) visitados
      | nodo == meta = [(nodo, costo)]
      | nodo `elem` visitados = buscar cola visitados
      | otherwise =
          let nuevos = [ (v, costo + c) | (v, c) <- sucesores nodo, v `notElem` visitados ]
              ordenados = sortOn (\(v, c) -> c + heur_fn v) (cola ++ nuevos)
          in (nodo, costo) : buscar ordenados (nodo : visitados)
```
